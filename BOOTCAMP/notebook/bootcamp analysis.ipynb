```python
df['Profit_NGN'] = (
    df['Revenue_Final_NGN'] -
    (df['Cost_Per_Unit_NGN'] * df['Units'])
)
```

```python
df['Profit_Margin_Pct'] = (
    df['Profit_NGN'] / df['Revenue_Final_NGN'] * 100
)
```

```python
state_performance = (
    df.groupby('State')
      .agg(
          Revenue=('Revenue_Final_NGN', 'sum'),
          Cost=('Cost_Per_Unit_NGN', lambda x: 0),
          Profit=('Profit_NGN', 'sum'),
          Units=('Units', 'sum'),
          Transactions=('Transaction_ID', 'count')
      )
      .sort_values('Profit', ascending=False)
)

state_performance
```

```python
state_performance = (
    df.groupby('State')
      .agg(
          Revenue=('Revenue_Final_NGN', 'sum'),
          Profit=('Profit_NGN', 'sum'),
          Units=('Units', 'sum'),
          Transactions=('Transaction_ID', 'count')
      )
)

state_performance['Profit_Margin_Pct'] = (
    state_performance['Profit'] /
    state_performance['Revenue'] * 100
)

state_performance.sort_values('Profit', ascending=False)
```

```python
category_performance = (
    df.groupby('Product_Category')
      .agg(
          Revenue=('Revenue_Final_NGN', 'sum'),
          Profit=('Profit_NGN', 'sum'),
          Units=('Units', 'sum'),
          Transactions=('Transaction_ID', 'count')
      )
)

category_performance['Profit_Margin_Pct'] = (
    category_performance['Profit'] /
    category_performance['Revenue'] * 100
)

category_performance.sort_values('Profit', ascending=False)
```

```python
df['Discount_Band'] = pd.cut(
    df['Discount_Pct'],
    bins=[-0.001, 0.05, 0.10, 0.15, 1],
    labels=['0-5%', '5-10%', '10-15%', '15%+']
)
```

```python
discount_analysis = (
    df.groupby('Discount_Band', observed=True)
      .agg(
          Transactions=('Transaction_ID', 'count'),
          Revenue=('Revenue_Final_NGN', 'sum'),
          Profit=('Profit_NGN', 'sum'),
          Units=('Units', 'sum')
      )
)

discount_analysis['Profit_Margin_Pct'] = (
    discount_analysis['Profit'] /
    discount_analysis['Revenue'] * 100
)

discount_analysis
```

```python
channel_return = (
    df.groupby('Channel')
      .agg(
          Transactions=('Transaction_ID', 'count'),
          Returns=('Return_Flag', 'sum')
      )
)

channel_return['Return_Rate_Pct'] = (
    channel_return['Returns'] /
    channel_return['Transactions'] * 100
)

channel_return.sort_values('Return_Rate_Pct')
```
